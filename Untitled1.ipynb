```python
import numpy as np
import matplotlib.pyplot as plt

# Create a blank 200x200 RGB image
img = np.zeros((200, 200, 3), dtype=np.uint8)

print("Image shape:", img.shape)
print("Image data type:", img.dtype)
print("Image size:", img.nbytes, "bytes")

plt.figure(figsize=(4, 4))
plt.imshow(img)
plt.title("Blank Image")
plt.axis("off")
plt.show()
```

```python
# Create a white background
img = np.full((200, 200, 3), 255, dtype=np.uint8)

# Red block: rows 10-59, columns 20-89
img[10:60, 20:90, 0] = 255
img[10:60, 20:90, 1] = 0
img[10:60, 20:90, 2] = 0

# Green block: rows 70-119, columns 20-89
img[70:120, 20:90] = [0, 255, 0]

# Blue block: rows 130-179, columns 20-89
img[130:180, 20:90] = [0, 0, 255]

# Draw horizontal black stripes
img[30:35, :, :] = 0
img[90:95, :, :] = 0
img[150:155, :, :] = 0

# Draw a yellow block
img[50:100, 120:180] = [255, 255, 0]

plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.title("Synthetic Image with Color Blocks and Stripes")
plt.axis("off")
plt.show()
```

```python
# 1. Crop a region of the image
cropped_img = img[20:120, 30:150, :]

# 2. Flip horizontally using negative-step slicing
horizontal_flip = img[:, ::-1, :]

# 3. Flip vertically
vertical_flip = img[::-1, :, :]

# 4. Extract individual RGB channels
red_channel = img[:, :, 0]
green_channel = img[:, :, 1]
blue_channel = img[:, :, 2]

# Display results
fig, axes = plt.subplots(2, 3, figsize=(12, 8))

axes[0, 0].imshow(img)
axes[0, 0].set_title("Original Image")

axes[0, 1].imshow(cropped_img)
axes[0, 1].set_title("Cropped Image")

axes[0, 2].imshow(horizontal_flip)
axes[0, 2].set_title("Horizontal Flip")

axes[1, 0].imshow(vertical_flip)
axes[1, 0].set_title("Vertical Flip")

axes[1, 1].imshow(red_channel, cmap="gray", vmin=0, vmax=255)
axes[1, 1].set_title("Red Channel")

axes[1, 2].imshow(green_channel, cmap="gray", vmin=0, vmax=255)
axes[1, 2].set_title("Green Channel")

for ax in axes.flat:
    ax.axis("off")

plt.tight_layout()
plt.show()

# Display the blue channel separately
plt.figure(figsize=(4, 4))
plt.imshow(blue_channel, cmap="gray", vmin=0, vmax=255)
plt.title("Blue Channel")
plt.axis("off")
plt.show()

print("Original shape:", img.shape)
print("Cropped shape:", cropped_img.shape)
print("Red channel shape:", red_channel.shape)
```

```python
# Convert the image to float32
img_float = img.astype(np.float32)

# Increase brightness by 40
brighter_img = np.clip(img_float + 40, 0, 255).astype(np.uint8)

# Decrease brightness by 40
darker_img = np.clip(img_float - 40, 0, 255).astype(np.uint8)

# Create a mask for bright pixels
# A pixel is considered bright if its maximum RGB value is above 200
bright_mask = np.max(img, axis=2) > 200

# Create an image showing only the bright pixels
bright_pixels_img = np.zeros_like(img)
bright_pixels_img[bright_mask] = img[bright_mask]

# Display the images
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].imshow(img)
axes[0, 0].set_title("Original Image")

axes[0, 1].imshow(brighter_img)
axes[0, 1].set_title("Brighter Image (+40)")

axes[1, 0].imshow(darker_img)
axes[1, 0].set_title("Darker Image (-40)")

axes[1, 1].imshow(bright_pixels_img)
axes[1, 1].set_title("Bright Pixels Only")

for ax in axes.flat:
    ax.axis("off")

plt.tight_layout()
plt.show()

print("Original data type:", img.dtype)
print("Float data type:", img_float.dtype)
print("Bright pixel count:", np.count_nonzero(bright_mask))
print("Total pixels:", img.shape[0] * img.shape[1])
```

```python
# Convert the image to different data types
img_uint8 = img.astype(np.uint8)
img_float32 = img.astype(np.float32)
img_float64 = img.astype(np.float64)

# Compare memory usage
print("Memory Comparison")
print("-----------------")
print("uint8  :", img_uint8.nbytes, "bytes")
print("float32:", img_float32.nbytes, "bytes")
print("float64:", img_float64.nbytes, "bytes")

print("\nMemory in KB")
print("------------")
print("uint8  :", img_uint8.nbytes / 1024, "KB")
print("float32:", img_float32.nbytes / 1024, "KB")
print("float64:", img_float64.nbytes / 1024, "KB")

# Display uint8 and float64 images
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

axes[0].imshow(img_uint8)
axes[0].set_title("uint8 Image")

axes[1].imshow(img_float64 / 255.0)
axes[1].set_title("float64 Image (Normalized)")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()

# Calculate memory savings
saving = (1 - img_uint8.nbytes / img_float64.nbytes) * 100
print(f"\nMemory saved using uint8: {saving:.2f}%")
```
